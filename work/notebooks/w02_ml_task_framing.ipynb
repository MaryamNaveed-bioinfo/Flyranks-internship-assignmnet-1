## 1. My lane as an ML task (type)

*Classification, clustering, ranking, or scoring — which one, and why?*

**Task type: Classification (with ranking/scoring on top)**

My lane predicts whether a page is likely declining - a binary classification problem
(declining vs. not declining). But the final output isn't just a yes/no label: I convert
the model's predicted probability into a ranking score, so pages can be sorted from
"most worth reviewing" to "least worth reviewing." This mirrors what I already practiced
in the starter notebook, where a decision tree predicted `is_declining_label` and its
probability output was used to rank pages by Precision@K.

## 2. Target or proxy

*What would you predict? Where does that label come from — observed outcome or a defined rule?*

**Target: `is_declining_label`, derived from `trend_direction == "down"`**

This is a *proxy* label, not a perfect one — it reflects the current trend bucket rather
than a guaranteed future outcome. A stronger version (which I may build toward later)
would predict decline over a future window (e.g. "will this page's traffic drop in the
next 30 days"), using only features known before that window starts. For now, this proxy
is a reasonable starting point since it's directly available in the starter data and
mirrors the pipeline I already tested.

## 3. Success metric

*One metric you can defend. What number means 'good'?*

**Success metric: Precision@50**

Since a reviewer only has time to check a limited number of pages, what matters most is:
"out of the top 50 pages my model flags, how many are actually declining?" This is more
defensible than overall accuracy, which could look artificially high while still being
useless for prioritization. In the starter notebook, a random forest scored 0.740
Precision@50 , meaning about 37 of the top 50 flagged pages were genuinely declining,
a concrete, actionable number for a reviewer with limited time.

## 5. Why ML beats a fixed rule here

*What makes the pattern too messy for an if-statement?*

A simple hand-written rule (like "flag pages that are stale AND still getting traffic")
is a reasonable starting point, and I tested exactly this in the starter notebook it
scored well at the very top of the ranking (Precision@20: 0.900). But its performance
dropped further down the list (Precision@50: 0.680), because a fixed rule can only combine
a small number of conditions with a human-chosen threshold.

A model can weigh many more signals simultaneously (traffic, position, age, freshness,
engagement, content type) and find combinations a person wouldn't think to hand-code
which is exactly why the random forest reached 0.740 Precision@50, beating the hand rule
at that depth. ML doesn't replace human judgment here; it extends how many pages a
reviewer can meaningfully prioritize before running out of simple rules to apply.
