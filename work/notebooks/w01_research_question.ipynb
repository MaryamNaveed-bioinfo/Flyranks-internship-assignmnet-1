## 2. The question: decision, action, cost of a wrong call

Output: A ranked list of pages, ordered by priority score highest-priority pages
at the top.

Action: A reviewer could refresh, expand, protect, prune, or simply monitor a flagged page, depending on the reason code attached to it.

is taken.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd
df = pd.read_csv("data/raw/content_refresh_anonymized.csv")

# Number 1: how common is decline?
declining_rate = df["trend_direction"].str.lower().eq("down").mean()
print(f"{declining_rate:.1%} of pages are currently declining")

# Number 2: how many pages are stale but still get real traffic?
stale_visible = df[(df["days_since_last_update"] >= 180) & (df["impressions_90d"] >= 500)]
print(f"{len(stale_visible):,} pages are stale (180+ days unedited) but still get 500+ impressions")

# Number 3: how much traffic is tied up in declining pages?
declining_impressions = df[df["trend_direction"].str.lower() == "down"]["impressions_90d"].sum()
print(f"Declining pages account for {declining_impressions:,.0f} total impressions over 90 days")
print(df["impressions_90d"].describe())
```

These numbers show that decline is not a rare edge case: 54.2% of pages in the starter
dataset are currently declining. A smaller set of 17 pages are both stale (unedited for
180+ days) and still receiving significant traffic (500+ impressions) — clear candidates
for review. Declining pages collectively account for roughly 80 million impressions over
90 days, though this total is concentrated in a smaller number of high-traffic pages
(the median page gets only 731 impressions, while the top pages get over 500,000) — so a
review process should weight both how many pages are affected and how much traffic is
actually at stake per page, not treat all declining pages as equal priority.
